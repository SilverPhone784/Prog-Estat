```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

# Funções necessárias:

**Gama**

```python
def f_gama(x, alpha, beta):
  return (beta**alpha)/(math.gamma(alpha)) * x**(alpha-1) * np.exp(-beta * x)

def g_gama(x, lam):
  return lam*np.exp(-lam*x)

def gama(alpha, beta, M):
  lam = beta/alpha
  c = (beta**alpha)/(math.gamma(alpha) * lam) * ((alpha-1)/(beta-lam))**(alpha-1) * np.exp(-(alpha-1))
  amostra = []

  while len(amostra) < M:
    u = np.random.uniform()
    y = (-1/lam) * np.log(np.random.uniform())

    if(u < f_gama(y, alpha, beta)/(c*g_gama(y, lam))):
      amostra.append(y)

  return amostra
```

---

**Beta**

```python
def f_beta(x, alpha, beta):
  return math.gamma(alpha + beta)/(math.gamma(alpha)*math.gamma(beta)) * x**(alpha-1) * (1-x)**(beta-1)

def beta(alpha, beta, M):
    valor = []
    for x in np.linspace(0, 1, M):
        valor.append(f_beta(x, alpha, beta))

    c = np.max(valor)+0.05

    amostra = []

    while len(amostra) < M:
        u = np.random.uniform()
        y = np.random.uniform()

        if(u < f_beta(y, alpha, beta)/c):
            amostra.append(y)

    return amostra
```

---

**Normal**

```python
def f_normal(x):
  return 2*np.exp(-(x**2)/2)/np.sqrt(2*np.pi)

def g_normal(x):
  return np.exp(-x)

def normal(mi, sigma, M):
  c = 2*np.exp(1/2)/np.sqrt(2*np.pi)

  amostra = []
  while len(amostra) < M:
    u = np.random.uniform()
    y = (-1) * np.log(np.random.uniform())

    if(u < f_normal(y)/(c*g_normal(y))):
      amostra.append(y)

  for i in range(len(amostra)):
    if(np.random.uniform() < 0.5):
      amostra[i] = -amostra[i]

  if(mi != 0 or sigma != 1):
    for i in range(len(amostra)):
      amostra[i] = (amostra[i] * sigma) + mi

  return amostra
```

---

**Exponencial**

```python
def exponencial(lam, M):
  amostra = []
  for i in range(M):
    x = (-1/lam) * np.log(np.random.uniform())
    amostra.append(x)

  return amostra
```

# Exercício 1

Considere $X\sim\operatorname{Gamma}(3,2)$. Estime por Monte Carlo

$$
\mu=E[\sqrt{X}].
$$

Calcule a variância do estimador e construa um intervalo de confiança de 95%. Compare com o valor teórico.

**Valor teórico:**

$$
\mu=\frac{\Gamma(7/2)}{\Gamma(3)\sqrt{2}}\approx 1{,}174982.
$$


```python
n = 10_000
dados = gama(3, 2, n)

mi_chapeu = np.mean(np.sqrt(dados))
sigma_chapeu = np.std(np.sqrt(dados))

ic = [mi_chapeu - 1.96*sigma_chapeu/np.sqrt(n), mi_chapeu + 1.96*sigma_chapeu/np.sqrt(n)]

print(f"{mi_chapeu} média estimada")
print(f"({ic[0]} ; {ic[1]}) I.C. 95%")
```

# Exercício 7

Considere $X_1,\ldots,X_5$ independentes, com $X_i\sim\operatorname{Exp}(2)$. Defina

$$
M=\max(X_1,\ldots,X_5).
$$

Estime $E[M]$ por Monte Carlo. Calcule a variância do estimador e construa um intervalo de confiança de 95%.

**Valor teórico:**

$$
E[M]=\frac{1}{2}\sum_{k=1}^{5}\frac{1}{k}\approx 1{,}141667.

```python
X1 = exponencial(2, 10_000)
X2 = exponencial(2, 10_000)
X3 = exponencial(2, 10_000)
X4 = exponencial(2, 10_000)
X5 = exponencial(2, 10_000)

amostra = []
for i in range(10_000):
  amostra.append(np.max([X1[i], X2[i], X3[i], X4[i], X5[i]]))

max_chapeu = np.mean(amostra)
sigma_chapeu = np.std(amostra)

ic = [max_chapeu - 1.96*sigma_chapeu/np.sqrt(10_000), max_chapeu + 1.96*sigma_chapeu/np.sqrt(10_000)]

print(f"{max_chapeu} média do maximo estimada")
print(f"({ic[0]} ; {ic[1]}) I.C. 95%")
```
