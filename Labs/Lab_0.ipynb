## Passo 1 – Criar o dicionário principal
As chaves devem ser nomes de distribuições: `"Normal"`, `"Exponencial"`, `"Uniforme"`  
Os valores devem ser dicionários contendo:  
- `"params"` → parâmetros da distribuição  
- `"dados"` → inicialmente `None`  
- `"estatisticas"` → inicialmente `None`  

Exemplo de estrutura:

    distribuicoes = {
        "Normal": {"params": {"loc": 0, "scale": 1}, "dados": None, "estatisticas": None},
        "Exponencial": {"params": {"scale": 2}, "dados": None, "estatisticas": None},
        "Uniforme": {"params": {"low": -1, "high": 1}, "dados": None, "estatisticas": None}
    }

---

```python
distribuicoes = {
    "Normal": {"params": {"loc": 0, "scale": 1}, "dados": None, "estatisticas": None},
    "Exponencial": {"params": {"scale": 2}, "dados": None, "estatisticas": None},
    "Uniforme": {"params": {"low": -1, "high": 1}, "dados": None, "estatisticas": None}
}
```

```python
distribuicoes["Normal"]["params"]
```

## Passo 2 – Importar bibliotecas
Importe as bibliotecas:  
- `numpy` para simulação  
- `pandas` para organização dos resultados  
- `matplotlib` para visualização  

---

## Passo 3 – Gerar os dados
Para cada distribuição, gere uma amostra de tamanho **1000** usando `numpy.random` e salve em `"dados"`.  

---


```python
# Parte de Importar as bibliotecas
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(847063) #Deixei o meu RA de seed para os resultados não mudarem a cada vez que eu rodar o código

#Gerando os dados e armazendo-os no dicionario de cada distribuição
distribuicoes["Normal"]["dados"] = np.random.normal(loc=0, scale=1, size=1000)
distribuicoes["Exponencial"]["dados"] = np.random.exponential(scale=2, size=1000)
distribuicoes["Uniforme"]["dados"] = np.random.uniform(low=-1, high=1, size=1000)
```

## Passo 4 – Calcular estatísticas básicas
Para cada distribuição, calcule:  
- Média  
- Mediana  
- Variância  
- Desvio padrão  
Armazene em um dicionário dentro da chave `"estatisticas"`.  

---

```python
#Atualizando as estatisticas da normal a partir dos dados gerados
distribuicoes["Normal"]["estatisticas"] = {"média": np.mean(distribuicoes["Normal"]["dados"]),
                                           "mediana": np.median(distribuicoes["Normal"]["dados"]),
                                           "desvio padrão": np.std(distribuicoes["Normal"]["dados"]),
                                           "variância": np.var(distribuicoes["Normal"]["dados"])}
distribuicoes["Normal"]["estatisticas"]
```

```python
#Atualizando as estatisticas da exponencial a partir dos dados gerados
distribuicoes["Exponencial"]["estatisticas"] = {"média": np.mean(distribuicoes["Exponencial"]["dados"]),
                                           "mediana": np.median(distribuicoes["Exponencial"]["dados"]),
                                           "desvio padrão": np.std(distribuicoes["Exponencial"]["dados"]),
                                           "variância": np.var(distribuicoes["Exponencial"]["dados"])}
distribuicoes["Exponencial"]["estatisticas"]
```

```python
#Atualizando as estatisticas da uniforme a partir dos dados gerados
distribuicoes["Uniforme"]["estatisticas"] = {"média": np.mean(distribuicoes["Uniforme"]["dados"]),
                                           "mediana": np.median(distribuicoes["Uniforme"]["dados"]),
                                           "desvio padrão": np.std(distribuicoes["Uniforme"]["dados"]),
                                           "variância": np.var(distribuicoes["Uniforme"]["dados"])}
distribuicoes["Uniforme"]["estatisticas"]
```

## Passo 5 – Visualizar os dados
Para cada distribuição, faça um **histograma** mostrando a forma da distribuição.  

---


```python
plt.hist(distribuicoes["Normal"]["dados"], density=True)
plt.title("Histograma da Normal")
plt.show()
```

```python
plt.hist(distribuicoes["Exponencial"]["dados"], density=True)
plt.title("Histograma da Exponencial")
plt.show()
```

```python
plt.hist(distribuicoes["Uniforme"]["dados"], density=True)
plt.title("Histograma da Uniforme")
plt.show()
```

## Passo 6 – Organizar resultados em Pandas
Crie um `DataFrame` contendo as estatísticas de todas as distribuições.  

---


```python
df = pd.DataFrame({"Normal": distribuicoes["Normal"]["estatisticas"]}) #Organizando o dataframe, com as linhas sendo as Estatisticas e a coluna a distribuição
df
```

```python
df.insert(column="Exponencial", loc=1, value=distribuicoes["Exponencial"]["estatisticas"]) #Inserindo a coluna da exponencial
df.insert(column="Uniforme", loc=2, value=distribuicoes["Uniforme"]["estatisticas"]) #Inserindo a coluna da uniforme
df
```


## Passo 7 – Adicionar novas distribuições
- Inclua mais uma distribuição, como `"Poisson"` ou `"Binomial"`.  
- Gere dados usando `numpy.random`.  
- Calcule as estatísticas básicas e atualize o `DataFrame`.  

---

```python
distribuicoes.update({"Poisson": {"params": {"lam": 2}, "dados": np.random.poisson(lam=2, size=1000), "estatisticas": None }}) # Adicionando uma poisson com lambda = 2 ao dicionario de distribuições e gerando os dados
distribuicoes['Poisson']['estatisticas'] = {
    "média": np.mean(distribuicoes["Poisson"]["dados"]),
    "mediana": np.median(distribuicoes["Poisson"]["dados"]),
    "desvio padrão": np.std(distribuicoes["Poisson"]["dados"]),
    "variância": np.var(distribuicoes["Poisson"]["dados"])
} #Fazendo as estatísticas da Poisson a partir dos dados gerados
```

```python
plt.hist(distribuicoes["Poisson"]["dados"], density=True)
plt.title("Histograma da Poisson")
plt.show()
```

```python
df.insert(column="Poisson", loc=3, value=distribuicoes["Poisson"]["estatisticas"]) #Inserindo a Poisson no dataframe
df
```

## Passo 8 – Estatísticas adicionais
Para cada distribuição, calcule também:  
- Mínimo (`np.min`)  
- Máximo (`np.max`)  
- Quartis (`np.percentile`)  

Atualize o campo `"estatisticas"` e o `DataFrame` com essas novas colunas.  

---

```python
#Loop for para calcular as novas estatisticas de cada distribuição e adiciona-las ao dicionario estatisticas dentro do dicionario de cada distribuição
for key in distribuicoes.keys():
  distribuicoes[f"{key}"]["estatisticas"].update({"mínimo": np.min(distribuicoes[f"{key}"]["dados"]),
                                                "máximo": np.max(distribuicoes[f"{key}"]["dados"]),
                                                "q1": np.percentile(distribuicoes[f"{key}"]["dados"], 25),
                                                "q3": np.percentile(distribuicoes[f"{key}"]["dados"], 75)})
  print(key, distribuicoes[f"{key}"]["estatisticas"])
```

```python
estatisticas_novas = ['mínimo', 'máximo', 'q1', 'q3'] # Lista com as estatisticas novas
novas_linhas = pd.DataFrame({"Normal": {chave: distribuicoes["Normal"]["estatisticas"][chave] for chave in estatisticas_novas}}) # Criando um Dataframe para as linhas com estatisticas novas usando dict comprehension para selecionar apenas as estatisticas novas (se não fizer isso o concat duplica as linhas que já havia antes)
novas_linhas.insert(column="Exponencial", loc=1, value={chave: distribuicoes["Exponencial"]["estatisticas"][chave] for chave in estatisticas_novas}) # Inserindo as demais distribuições nesse dataframe usando a mesma ideia do dict comprehension
novas_linhas.insert(column="Uniforme", loc=2, value={chave: distribuicoes["Uniforme"]["estatisticas"][chave] for chave in estatisticas_novas})
novas_linhas.insert(column="Poisson", loc=3, value={chave: distribuicoes["Poisson"]["estatisticas"][chave] for chave in estatisticas_novas})
df = pd.concat([df, novas_linhas]) # Usando o concat para juntar tudo em um único dataframe
df
```

## Passo 9 – Comparar distribuições graficamente
Faça um **boxplot** com todas as distribuições no mesmo gráfico, com título e rótulos adequados.  

---

```python
import seaborn as sns # Importei o Seaborn para fazer o boxplot por ser mais fácil de usar e ser uma biblioteca que eu já trabalhei com na disciplina de Analise Descritiva e Exploratoria de Dados (no 2° semestre)

# Transformando os dados para uma forma simples de aplicar no seaborn
box = {
    "Normal": distribuicoes["Normal"]["dados"],
    "Exponencial": distribuicoes["Exponencial"]["dados"],
    "Uniforme": distribuicoes["Uniforme"]["dados"],
    "Poisson": distribuicoes["Poisson"]["dados"]
}

df_box = pd.DataFrame(box)

sns.boxplot(
    data=df_box
)
plt.grid(axis='y', linestyle='--', linewidth=1.2, color='gray', alpha=0.7)
plt.show()
```

## Passo 10 – Comparar estatísticas calculadas com valores teóricos
- Para cada distribuição, obtenha os valores teóricos (esperados) a partir dos parâmetros em `"params"`.  
- Compare esses valores com as estatísticas calculadas a partir da amostra.   
- Crie um gráfico de barras mostrando `media_teorica` e `media_amostra` para cada distribuição.  

---

```python
comparacao = {
    "Normal": {"Média Teorica": distribuicoes['Normal']['params']['loc'],
               "Média Amostral": distribuicoes['Normal']['estatisticas']['média'],
               "Variância Teórica": distribuicoes['Normal']['params']['scale'],
               "Variância Amostral": distribuicoes['Normal']['estatisticas']['variância']},
    "Exponencial": {"Média Teorica": distribuicoes['Exponencial']['params']['scale'],
                    "Média Amostral": distribuicoes['Exponencial']['estatisticas']['média']},
    "Uniforme": {"Média Teorica": (distribuicoes['Uniforme']['params']['low'] + distribuicoes['Uniforme']['params']['high']) / 2,
                 "Média Amostral": distribuicoes['Uniforme']['estatisticas']['média']},
    "Poisson": {"Média Teorica": distribuicoes['Poisson']['params']['lam'],
                "Média Amostral": distribuicoes['Poisson']['estatisticas']['média']}
}
comparacao
```

```python
plt.figure(figsize=(10, 6))
plt.title("Comparação com a Normal")
plt.bar(comparacao["Normal"].keys(), comparacao["Normal"].values())
plt.grid(axis='y', linestyle='--', linewidth=1.2, color='gray', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.title("Comparação com a Exponencial")
plt.bar(comparacao["Exponencial"].keys(), comparacao["Exponencial"].values())
plt.grid(axis='y', linestyle='--', linewidth=1.2, color='gray', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.title("Comparação com a Uniforme")
plt.bar(comparacao["Uniforme"].keys(), comparacao["Uniforme"].values())
plt.grid(axis='y', linestyle='--', linewidth=1.2, color='gray', alpha=0.7)
plt.ylim(0, 0.5)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.title("Comparação com a Poisson")
plt.bar(comparacao["Poisson"].keys(), comparacao["Poisson"].values())
plt.grid(axis='y', linestyle='--', linewidth=1.2, color='gray', alpha=0.7)
plt.show()
```
