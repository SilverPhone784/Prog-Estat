## Funções Necessarias

**Normal**

```python
def f_normal(x):
  return 2*np.exp(-(x**2)/2)/np.sqrt(2*np.pi)

def g_normal(x):
  return np.exp(-x)

def normal(mi, sigma, M):
  c = 2*np.exp(1/2)/np.sqrt(2*np.pi)

  amostra = []
  while len(amostra) < M:
    u = np.random.uniform()
    y = (-1) * np.log(np.random.uniform())

    if(u < f_normal(y)/(c*g_normal(y))):
      amostra.append(y)

  for i in range(len(amostra)):
    if(np.random.uniform() < 0.5):
      amostra[i] = -amostra[i]

  if(mi != 0 or sigma != 1):
    for i in range(len(amostra)):
      amostra[i] = (amostra[i] * sigma) + mi

  return amostra
```

---

**Exponencial**

```python
def exponencial(lam, M):
  amostra = []
  for i in range(M):
    x = (-1/lam) * np.log(np.random.uniform())
    amostra.append(x)

  return amostra
```

## Exercícios — Simulação de Monte Carlo

Em todos os exercícios, utilize $n=10^4$ simulações independentes, salvo indicação contrária.

```python
n = 10**4
```

### Exercício 1 — Estimação de integrais

Considere a integral

$$
I=\int_0^1 e^{-x^2}\,dx.
$$

Utilizando $X\sim U(0,1)$, estime $I$ por Monte Carlo, calcule a variância do estimador e construa um intervalo de confiança de 95%.


```python
u = np.random.uniform(size=n)

i_chapeu = np.mean(np.exp(-(u**2)))
sigma_chapeu = np.std(np.exp(-(u**2)))

ic = [mi_chapeu - (1.96*sigma_chapeu/np.sqrt(n)), mi_chapeu + (1.96*sigma_chapeu/np.sqrt(n))]

print(f"{i_chapeu} Integral estimada")
print(f"({ic[0]} ; {ic[1]}) I.C. 95%")
```

### Exercício 2 — Estimação de probabilidades

Considere $X\sim N(0,1)$ e a probabilidade

$$
p=P(X>1).
$$

Estime $p$ por Monte Carlo, calcule a variância do estimador e construa um intervalo de confiança de 95%.


```python
dados = np.array(normal(0, 1, n))

p_chapeu = np.mean(dados > 1)
sigma_chapeu = np.std(dados > 1)

ic = [p_chapeu - (1.96*sigma_chapeu/np.sqrt(n)) , p_chapeu + (1.96*sigma_chapeu/np.sqrt(n))]

print(f"{p_chapeu} proporção estimada")
print(f"({ic[0]} ; {ic[1]}) I.C. 95%")
```

### Exercício 3 — Estimação de $\pi$

Considere $(X,Y)$ uniformemente distribuído em $[0,1]^2$ e o evento

$$
A=\{X^2+Y^2\leq 1\}.
$$

Utilize Monte Carlo para estimar $\pi$, calcule a variância do estimador e construa um intervalo de confiança de 95%.

```python
X = np.random.uniform(size=n)
Y = np.random.uniform(size=n)

soma = np.array(X**2 + Y**2)

pi_chapeu = 4*np.mean(soma <= 1)
sigma_chapeu = 4*np.std(soma <= 1)

ic = [pi_chapeu - (1.96*sigma_chapeu/np.sqrt(n)), pi_chapeu + (1.96*sigma_chapeu/np.sqrt(n))]

print(f"{pi_chapeu} π estimado")
print(f"({ic[0]} ; {ic[1]}) I.C. 95%")
```

### Exercício 4 — Estimação de média e variância

Considere $X\sim U(0,1)$.

Estime $E[X]$ e $\operatorname{Var}(X)$ por Monte Carlo. Calcule a variância do estimador da média e construa um intervalo de confiança de 95% para $E[X]$.


```python
u = np.random.uniform(size=n)

mi_chapeu = np.mean(u)
sigma_chapeu = np.std(u)

ic = [mi_chapeu - 1.96*sigma_chapeu/np.sqrt(n), mi_chapeu + 1.96*sigma_chapeu/np.sqrt(n)]

print(f"{mi_chapeu} média estimada")
print(f"({ic[0]} ; {ic[1]}) I.C. 95% \n")

var_chapeu = np.var(u)
ic = [var_chapeu - 1.96*sigma_chapeu/np.sqrt(n), var_chapeu + 1.96*sigma_chapeu/np.sqrt(n)]

print(f"{var_chapeu} variancia estimada")
print(f"({ic[0]} ; {ic[1]}) I.C. 95%")
```

### Exercício 5 — Integração em duas dimensões

Considere a integral

$$
I=\int_0^1\int_0^1 e^{-(x^2+y^2)}\,dx\,dy.
$$

Utilizando $(X,Y)\sim U([0,1]^2)$, estime $I$ por Monte Carlo, calcule a variância do estimador e construa um intervalo de confiança de 95%.


```python
X = np.random.uniform(size=n)
Y = np.random.uniform(size=n)

i_chapeu = np.mean(np.exp(-(X**2 + Y**2)))
sigma_chapeu = np.std(np.exp(-(X**2 + Y**2)))

ic = [i_chapeu - 1.96*sigma_chapeu/np.sqrt(n), i_chapeu + 1.96*sigma_chapeu/np.sqrt(n)]

print(f"{i_chapeu} integral estimada")
print(f"({ic[0]} ; {ic[1]}) I.C. 95%")
```

### Exercício 6 — Esperança de uma função

Considere $X\sim \operatorname{Exp}(1)$ e

$$
\mu=E[e^{-X}].
$$

Estime $\mu$ por Monte Carlo, calcule a variância do estimador e construa um intervalo de confiança de 95%. Compare com o valor teórico.


```python
dados = exponencial(1, n)

mi_chapeu = np.mean(np.exp(-np.array(dados)))
sigma_chapeu = np.std(np.exp(-(np.array(dados))))

ic = [mi_chapeu - 1.96*sigma_chapeu/np.sqrt(n), mi_chapeu + 1.96*sigma_chapeu/np.sqrt(n)]

print(f"{mi_chapeu} média estimada")
print(f"({ic[0]} ; {ic[1]}) I.C. 95%")
```

### Exercício 7 — Integral imprópria

Considere a integral

$$
I=\int_0^\infty \frac{e^{-x}}{1+x}\,dx.
$$

Utilizando $X\sim \operatorname{Exp}(1)$, estime $I$ por Monte Carlo, calcule a variância do estimador e construa um intervalo de confiança de 95%.


```python
dados = exponencial(1, n)

i_chapeu = np.mean(1/(1 + np.array(dados)))
sigma_chapeu = np.std(1/(1 + np.array(dados)))

ic = [i_chapeu - 1.96*sigma_chapeu/np.sqrt(n), i_chapeu + 1.96*sigma_chapeu/np.sqrt(n)]

print(f"{i_chapeu} integral estimada")
print(f"({ic[0]} ; {ic[1]}) I.C. 95%")
```

### Exercício 8 — Integral oscilatória

Considere a integral truncada

$$
I_L=\int_0^L \frac{\sin(x)}{x}\,dx,
\qquad L=20.
$$

Utilizando $X\sim U(0,L)$, estime $I_L$ por Monte Carlo, calcule a variância do estimador e construa um intervalo de confiança de 95%. Compare com $\pi/2$.


```python
dados = np.random.uniform(size=n)*20

i_chapeu = np.mean(np.sin(dados)/dados)
sigma_chapeu = np.std(np.sin(dados)/dados)

ic = [i_chapeu - 1.96*sigma_chapeu/np.sqrt(n), i_chapeu + 1.96*sigma_chapeu/np.sqrt(n)]

print(f"{i_chapeu} integral estimada")
print(f"({ic[0]} ; {ic[1]}) I.C. 95%")
```
